```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

df = pd.read_csv('seasonal_agriculture_performance_dataset.csv')

df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())

sns.set_theme(style="whitegrid")
custom_green = '#75b59d'
sns.set_palette("Set2")
```

```python
plt.figure(figsize=(6, 5))
sns.boxplot(data=df, x='Season', y='Yield_Tonnes_Ha', color=custom_green, width=0.5)
plt.title('Yield (Tonnes/Ha) by Season', fontweight='bold')
plt.savefig('yield_by_season.png', transparent=True, bbox_inches='tight')
plt.show()

plt.figure(figsize=(6, 5))
sns.boxplot(data=df, x='Season', y='Profit_INR', color=custom_green, width=0.5)
plt.title('Profit (INR) by Season', fontweight='bold')
plt.savefig('profit_by_season.png', transparent=True, bbox_inches='tight')
plt.show()

plt.figure(figsize=(6, 5))
sns.barplot(data=df, x='Season', y='Rainfall_mm', color=custom_green, errorbar=None)
plt.title('Average Rainfall (mm) by Season', fontweight='bold')
plt.savefig('rainfall_by_season.png', transparent=True, bbox_inches='tight')
plt.show()

plt.figure(figsize=(6, 5))
sns.barplot(data=df, x='Season', y='Disease_Pest_Risk_pct', color=custom_green, errorbar=None)
plt.title('Average Disease/Pest Risk (%) by Season', fontweight='bold')
plt.savefig('disease_risk_by_season.png', transparent=True, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.barplot(data=df, x='Season', y='Fertilizer_kg_ha', color=custom_green, errorbar='sd')
plt.title('Average Fertilizer_kg_ha by Season', fontweight='bold')
plt.savefig('fertilizer_by_season.png', transparent=True, bbox_inches='tight')
plt.show()

plt.figure(figsize=(6, 5))
sns.barplot(data=df, x='Season', y='Pesticide_Litre_ha', color=custom_green, errorbar='sd')
plt.title('Average Pesticide_Litre_ha by Season', fontweight='bold')
plt.savefig('pesticide_by_season.png', transparent=True, bbox_inches='tight')
plt.show()

plt.figure(figsize=(6, 5))
sns.barplot(data=df, x='Season', y='Water_Used_m3', color=custom_green, errorbar='sd')
plt.title('Average Water_Used_m3 by Season', fontweight='bold')
plt.savefig('water_used_by_season.png', transparent=True, bbox_inches='tight')
plt.show()

plt.figure(figsize=(6, 5))
sns.barplot(data=df, x='Season', y='Water_Efficiency_t_per_1000m3', color=custom_green, errorbar='sd')
plt.title('Average Water Efficiency by Season', fontweight='bold')
plt.savefig('water_efficiency_by_season.png', transparent=True, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Season', hue='Irrigation_Method', palette='Set2')
plt.title('Irrigation Method Usage by Season', fontweight='bold')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.savefig('irrigation_usage.png', transparent=True, bbox_inches='tight')
plt.show()

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Irrigation_Method', y='Water_Efficiency_t_per_1000m3', hue='Season', palette='Set2', errorbar=None)
plt.title('Water Efficiency by Irrigation Method & Season', fontweight='bold')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.savefig('irrigation_efficiency.png', transparent=True, bbox_inches='tight')
plt.show()
```

```python
numeric_cols = [
    'Rainfall_mm', 'Avg_Temperature_C', 'Humidity_pct', 'Sunlight_Hours_Day',
    'Soil_pH', 'Soil_Moisture_pct', 'Nitrogen_kg_ha', 'Phosphorus_kg_ha',
    'Potassium_kg_ha', 'Fertilizer_kg_ha', 'Pesticide_Litre_ha', 'Seed_Quality_Score',
    'Yield_Tonnes_Ha', 'Water_Efficiency_t_per_1000m3', 'Disease_Pest_Risk_pct', 'Profit_INR'
]

plt.figure(figsize=(10, 8))
corr = df[numeric_cols].corr()

sns.heatmap(corr, annot=False, cmap='coolwarm', center=0, square=True, linewidths=.5)
plt.title('Correlation Matrix - Environmental, Input and Outcome Variables', fontweight='bold', pad=20)
plt.savefig('correlation_heatmap.png', transparent=True, bbox_inches='tight')
plt.show()
```

```python
econ_df = df.groupby('Season')[['Revenue_INR', 'Total_Cost_INR', 'Profit_INR']].mean().reset_index()
econ_melted = econ_df.melt(id_vars='Season', var_name='Metric', value_name='Amount')

plt.figure(figsize=(8, 6))
sns.barplot(data=econ_melted, x='Season', y='Amount', hue='Metric', palette='Set2')
plt.title('Average Revenue, Cost & Profit by Season', fontweight='bold')
plt.ylabel('Amount (INR)')
plt.savefig('economic_averages.png', transparent=True, bbox_inches='tight')
plt.show()

def calc_loss_pct(x):
    return (x['Profit_INR'] < 0).mean() * 100

loss_df = df.groupby('Season').apply(calc_loss_pct).reset_index(name='Loss_Pct')

plt.figure(figsize=(6, 6))
sns.barplot(data=loss_df, x='Season', y='Loss_Pct', color=custom_green)
plt.title('% of Loss-Making Farms by Season', fontweight='bold')
plt.ylabel('% Farms with Negative Profit')
plt.savefig('loss_making_pct.png', transparent=True, bbox_inches='tight')
plt.show()
```
